```python
# Plot the lines
plt.figure(figsize=(12, 6))
# Indices to highlight
highlight_indices = [1, 4, -1]  # 2nd, 5th, and last points (0-based indexing)

def plot_with_highlights(x, y, label, marker, color, highlight_indices=[1,4,-1]):
    plt.plot(x, y, label=label, marker=marker, color=color)
    # Highlight specific markers
    for idx in highlight_indices:
        plt.plot(
            x[idx], y[idx],  # Coordinates of the point
            marker=marker,  # Same marker
            color=color,  # Same color
            markersize=6,  # Larger marker size
            markeredgewidth=1,  # Thicker edge
            markeredgecolor='red',  # Black edge for emphasis
        )
plot_with_highlights(x, ap50, 'Full Depth Elastic Mask2Former', 'o', 'orange')
plot_with_highlights(x, elastic_backbone_ckpt_ap50, 'Resnet Depth Elastic Mask2Former', 's', 'blue')
plot_with_highlights(x, base_ckpt_ap50, 'Facebook Mask2Former ckpt', '^', 'black')
plot_with_highlights(x, further_finetuned_fb_ckpt_ap50, 'Finetuned Facebook Mask2Former ckpt', '^', 'purple')


# Add labels and title
plt.xlabel('Number of Parameter (millions)')
plt.ylabel('AP50')
plt.title('WS Mask2Former COCO AP50')

# Add a legend outside the graph
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))

# Adjust layout to make room for the legend
plt.tight_layout()

# Show the plot
plt.show()
```
